# Joint Probability :- What is the probability that an order is both delayed and receives a low customer satisfaction rating?

```python
# Joint Probability
joint_prob = len(
    master_df[
        (master_df["Delivery_Status"] == "Delayed") &
        (master_df["Satisfaction"] == "Low")
    ]
) / len(master_df)

print("Joint Probability (Delayed AND Low Satisfaction):", joint_prob)
```

Interpretation:
There is a 17.12% probability that an order is both delayed and receives a low satisfaction rating. This indicates that delivery delays are often associated with negative customer experiences and may reduce overall customer satisfaction.

Business Insight:
Reducing delivery delays can help decrease dissatisfied customers and improve customer retention.

# Conditional Probability
Business Question: What is the probability that a customer is satisfied given that the order was delivered on time?

```python
# Conditional Probability
on_time = master_df[master_df["Delivery_Status"] == "On Time"]

conditional_prob = len(
    on_time[on_time["Satisfaction"] == "High"]
) / len(on_time)

print("P(High Satisfaction | On Time Delivery):", conditional_prob)
```

Interpretation:
There is a 77.69% probability that a customer is highly satisfied when the order is delivered on time. This shows that timely delivery has a strong positive influence on customer satisfaction.

Business Insight:
Improving on-time delivery performance can significantly increase customer satisfaction and encourage repeat purchases.

# Two Sample Z-Test ->I used a Two-Sample Z-Test because I wanted to compare the average Final Price of two independent customer segments (New and Loyal).
 Both groups contain a very large number of observations (926,548 and 271,638), satisfying the large-sample requirement for the Z-Test.

```python
# Scenario: Is there a significant difference in the average Final Price between New and Loyal customers?
Null Hypothesis (H0):There is no significant difference in the average Final Price between New and Loyal customers.
                    H0:mu(New)=mu(Loyal)

Alternative Hypothesis (H1):There is a significant difference in the average Final Price between New and Loyal customers.
                            H1:mu(New)!=mu(Loyal)

```

```python
Identify Test Type:
- Not equal (≠) symbol → TWO-TAILED test
- Both N > 30 → Use Z-TEST
- Testing two independent groups
```

```python
# Librabry
from statsmodels.stats.weightstats import ztest
import numpy as np
from scipy.stats import norm
```

```python


# New Customers
new = master_df[master_df['customer_segment'] == 'New']['Final_Price']

# Loyal Customers
loyal = master_df[master_df['customer_segment'] == 'Loyal']['Final_Price']

# Mean
x1 = new.mean()
x2 = loyal.mean()

# Standard Deviation
sigma1 = new.std()
sigma2 = loyal.std()

# Sample Size
n1 = len(new)
n2 = len(loyal)

alpha = 0.05

# Standard Error
SE = np.sqrt((sigma1**2 / n1) + (sigma2**2 / n2))

# Z Statistic
z = (x1 - x2) / SE

# Two-Tailed P-value
p_value = 2 * (1 - norm.cdf(abs(z)))

# Critical Z-value
z_critical = norm.ppf(1 - alpha/2)

# Output
print("========== TWO SAMPLE Z-TEST ==========\n")

print("Mean (New Customers):", round(x1,2))
print("Mean (Loyal Customers):", round(x2,2))

print()

print("Standard Deviation (New):", round(sigma1,2))
print("Standard Deviation (Loyal):", round(sigma2,2))

print()

print("Sample Size (New):", n1)
print("Sample Size (Loyal):", n2)

print()

print("Standard Error:", round(SE,4))
print("Z Statistic:", round(z,4))
print("P Value:", round(p_value,6))
print("Critical Z Value:", round(z_critical,4))

print()

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: There is a significant difference in the Average Final Price between New and Loyal customers.")
else:
    print("Decision: Fail to Reject H0")
    print("Conclusion: There is no significant difference in the Average Final Price between New and Loyal customers.")
```

# Interpretation:
The Z-Test shows that there is a significant difference() in the average Final Price between New and Loyal customers.
New customers spend an average of Rs.6,452.52 per order, while Loyal customers spend Rs.6,271.22 per order. 
This means customer type has an impact on average order value.

# Business Recommendation:
The business should adopt different pricing, promotional, and retention strategies for each customer segment.
Personalized offers for New customers can improve conversion, while loyalty rewards and exclusive benefits 
can help retain Loyal customers and encourage repeat purchases.

```python
# Scenario: Is customer satisfaction associated with delivery status?

Null Hypothesis (H0): Customer Satisfaction and Delivery Status are independent.
Alternative Hypothesis (H₁):Customer Satisfaction and Delivery Status are associated.
```

```python
# Libary
import pandas as pd
from scipy.stats import chi2_contingency
```

```python
# creating Contingency Table
contingency_table = pd.crosstab(master_df['Delivery_Status'],
                                master_df['Satisfaction'])
print("***Count of customers in each category combination**\n")
print(contingency_table) 
```

```python
# Perform Chi-Square Test
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic =", round(chi2,2))
print("P Value =", p_value)
print("Degrees of Freedom =", dof)

alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("Customer Satisfaction is associated with Delivery Status.")
else:
    print("Fail to Reject H0")
    print("No significant association between Customer Satisfaction and Delivery Status.")
```

# Interpretation :
The Chi-Square Test shows that there is a significant association between Delivery Status and Customer Satisfaction. Since the p-value is much smaller than 0.05, we reject the null hypothesis and conclude that customer satisfaction depends on delivery status.

The contingency table shows that 1,052,331 delayed deliveries received high satisfaction, while 316,313 delayed deliveries received low satisfaction. Similarly, 372,141 on-time deliveries received high satisfaction, compared to 106,891 with low satisfaction. The Chi-Square statistic (127.74) and an extremely small p-value (1.277 × 10⁻²⁹) indicate a statistically significant relationship between delivery status and customer satisfaction.

# Business Interpretation:
Delivery performance has a measurable impact on customer satisfaction. Customers receiving on-time deliveries tend to have different satisfaction levels than those experiencing delayed deliveries. Therefore, improving delivery efficiency can help improve overall customer satisfaction.
