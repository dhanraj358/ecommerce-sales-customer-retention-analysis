```python
def dataset_overview(df, name):

    print("="*100)
    print(f"DATASET : {name}")
    print("="*100)

    print("\nShape")
    print(df.shape)

    print("\nHead")
    display(df.head())

    print("\nTail")
    display(df.tail())

    print("\nColumns")
    print(df.columns.tolist())

    print("\nData Types")
    print(df.dtypes)

    print("\nInfo")
    df.info()

    print("\nMissing Values")
    print(df.isnull().sum())

    print("\nDuplicate Rows")
    print(df.duplicated().sum())

    print("\nUnique Values")
    print(df.nunique())

    print("\nSummary Statistics")
    display(df.describe(include='all'))
```

```python
************************Feature Engineering***************
```

```python
#Purchase Conversion
orders_df["Purchase_Status"] = orders_df["order_status"].apply(
    lambda x: "Purchased" if x=="delivered" else "Not Purchased"
)
```

```python
# Repeat Purchase
orders_df = orders_df.sort_values(
    ["customer_id","order_purchase_timestamp"]
)

orders_df["Previous_Order"] = orders_df.groupby(
    "customer_id"
)["order_purchase_timestamp"].shift()

orders_df["Days_Between_Orders"] = (
orders_df["order_purchase_timestamp"]
-
orders_df["Previous_Order"]
).dt.days
```

```python
# Purchase Frequency
orders_df["Purchase_Frequency"] = orders_df.groupby(
"customer_id"
)["order_id"].transform("count")
```

```python
#Repeat Buyer
orders_df["Repeat_Buyer"] = np.where(
    orders_df["Purchase_Frequency"] > 1,
    "Yes",
    "No"
)
```

```python
# Total Price
order_items_df["Total_Price"] = (
    order_items_df["quantity"]
    * order_items_df["unit_price"]
)
```

```python
# Discount Amount
order_items_df["Discount_Amount"] = (
    order_items_df["Total_Price"]
    * order_items_df["discount(%)"]
    / 100
)
```

```python
#Final Price
order_items_df["Final_Price"] = (
    order_items_df["Total_Price"]
    - order_items_df["Discount_Amount"]
)
```

```python
# Estimated Profit
order_items_df["Estimated_Profit"] = (
    order_items_df["Final_Price"]
    - order_items_df["Product_Cost"]
    - order_items_df["shipping_cost"]
)
```

```python
#Shipping Category
order_items_df["Shipping_Type"] = pd.cut(
    order_items_df["shipping_cost"],
    bins=[0,20,50,1000],
    labels=["Low","Medium","High"]
)
```

```python
# Discount Band
order_items_df["Discount_Band"] = pd.cut(
order_items_df["discount(%)"],
bins=[0,10,20,30,100],
labels=["Low","Medium","High","Very High"]
)
```

```python
# Delivery_Days
orders_df["Delivery_Days"] = (
    orders_df["order_delivered_customer_date"] -
    orders_df["order_purchase_timestamp"]
).dt.days
```

```python
#Delivery Status
orders_df["Delivery_Status"] = (
orders_df["Delivery_Days"]
<=
(
orders_df["order_estimated_delivery_date"]
-
orders_df["order_purchase_timestamp"]
).dt.days
)

orders_df["Delivery_Status"] = orders_df[
"Delivery_Status"
].map(
{
True:"On Time",
False:"Delayed"
}
)
```

```python
#Satisfaction Level
reviews_df["Satisfaction"] = reviews_df["review_score"].apply(
lambda x:
"High"
if x>=4
else "Low"
)
```

```python
*******************Merging  all the Datasets********************
```

```python
# Orders + Customers
master_df1 = pd.merge(orders_df, customers_df, on="customer_id", how="inner")

# Add Order Items
master_df2 = pd.merge(master_df1, order_items_df, on="order_id", how="inner")

# Add Products
master_df3 = pd.merge(master_df2, products_df, on="product_id", how="inner")

# Add Reviews
master_df4 = pd.merge(master_df3, reviews_df, on="order_id", how="inner")

# Add Geolocation
master_df = pd.merge(master_df4, geolocation_df,
                     left_on="customer_zip_code",
                     right_on="geolocation_zip_code",
                     how="inner")
```
